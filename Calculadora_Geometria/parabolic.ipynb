```python
# imports
import math
```

```python
D = 1.3 # m
H = 0.2 # m
freq = 1.7e9 # Hz

lmbd = 3e8/freq
f = (D**2)/(16*H)
A = 0.6*(math.pi)*(D**2)/4
R = f/D
print(f"Foco: {f} m")
print(f"Area efetiva: {A} m^2")
print(f"Razão: {R}")
if R < 0.35:
    print("O sistema tem uma razão muito baixa")
elif R > 0.7:
    print("O sistema tem uma razão muito alta")
else:
    print("O sistema possui uma razão aceitável")

theta = math.atan2((0.5*D),f-H)
print(f"Máxima iluminação do receptor: {theta*180/math.pi} º")

G = 10*math.log10(6*((D/lmbd)**2))
print(f"Ganho: {G} dB")
```

```python
# Geometria
import numpy as np
import matplotlib.pyplot as plt

alpha = H/((D/2)**2)
x_vals = np.linspace(-D/2,D/2,2000)
parabolic_format = lambda x : alpha*(x**2)
y_vals = parabolic_format(x_vals)

plt.plot(x_vals,y_vals)
plt.scatter([0],[f])
```
